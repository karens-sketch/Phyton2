```python
#Ejercicio 1: Rectas
print("De acuerdo con la ecuación de una recta de la forma y=mx+b, donde m=pendiente y b=ordenada al origen")
m1 = float(input("Ingrese la pendiente de la recta 1: "))
b1 = float(input("Ingrese la ordenada al origen de la recta 1: "))
m2 = float(input("Ingrese la pendiente de la recta 2: "))
b2 = float(input("Ingrese la ordenada al origen de la recta 2: "))
if m1!=m2:  #Pendientes distintas
  print("Las rectas se intersectan")
elif m1==m2 and b1!=b2:   #pendiente igual y ordenadas distintas
  print("Las rectas son paralelas")
else:
  print("Las rectas son coincidentes")
```

```python
#Ejercicio 3: Desigualdad triangular y fórmula de Herón
while True:
  a=int(input("Ingresa el primer lado del triángulo"))
  b=int(input("Ingresa el segundo lado del triángulo"))
  c=int(input("Ingresa el tercer lado del triángulo"))
  s= (a+b+c)/2
  area = (s*(s-a)*(s-b)*(s-c))**(1/2)
  if a+b>c and a+c>b and b+c>a:
    if a == b and b == c:
      print(f"El triángulo es equilátero y su área es {area}")
    elif a == b or a==c or b==c:
      print(f"El triángulo es isosceles y su área es {area}")
    else:
      print(f"El triángulo es escaleno y su área es {area}")
    break
  else:
    print("Los lados no forman un triángulo")
```

```python
#Ejercicio 9: Volumen y Área de la Superficie de una Pirámide Regular Recta.
import math
h=int(input("Ingrese una altura mayor a 0"))
n=int(input("Ingrese el numeros de lados de la base (mayor a 3)"))
s=int(input("Ingrese la logitud del lado mayor a 0"))
if h<0 or n<=3 or s<0:
  print("ERROR. Ingrese un número válido")
else:
  a=(1/2)*s*(1/math.tan(math.pi/n))
  A=(1/2)*n*s*a
  l = math.sqrt((h**2)+(a**2))
  volumen = (1/3)*A*h
  asp = A+(1/2)*n*s*l
  print("El volumen de la pirámide rectangular es: ", volumen)
  print("El area de la pirámide rectangular es: ", asp)
```
