```python
#Ciclos while y for
cuentas = 0
while cuentas < 6:
    print(cuentas)
    cuentas = cuentas + 1
```

```python
x = 0
while not x == 5:
    x = x + 1
    print("x = %d" % x)
```

```python
#sentencial else
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```

```python
# Generar una secuencia de Fibonacci
a, b = 0, 1
while b < 100:
    print(b)
    a, b = b, a + b
```

```python
#FOR "Para todo elemento de una serie hacer una acción"
for i in range(0, 16, 3): #secuencia de números, que va de 0 a 10, de 2 en 2.
    print(i)
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
#Sentencia break (salir de bucles)
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1    # i+=1 -> i=1+1
```

```python
#Sentencia continue (filtrar datos)
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

```python
#Calcular el factorial de un numero dado
import math
x = int(input("Ingresa un numero"))
factorial = math.factorial(x)
for i in range(1, x+1):
  print(f"El factorial de {x}, es {factorial}")
```
