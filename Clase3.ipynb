```python
#Ejercicio superficie de la Tierra
import math
a= 6378137.0
b= 6356752.314245
e2=1-(b**(2)/a**(2))
e=math.sqrt(e2)
elipsoide = (2*math.pi*(a**2))*(1+((1-e**2)/e))*math.atanh(e)
print("El área de superficie de un esferoide oblato:", elipsoide)
r = 6371000
esfera = 4*math.pi*(r**2)
print("El área de la superficie de la Tierra:", esfera)
dif= ((esfera-elipsoide)/esfera)*100
print(f"La diferencia es del orden de {dif:.10f}%")
```
